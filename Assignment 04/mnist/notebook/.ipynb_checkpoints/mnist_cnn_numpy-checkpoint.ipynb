# MNIST CNN từ đầu với NumPy (From Scratch)

Xây dựng CNN hoàn toàn bằng NumPy, không dùng framework deep learning.

**Kiến trúc:** Conv2D → ReLU → MaxPool → Flatten → Dense → ReLU → Dense → Softmax

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os, time, struct, pickle
from pathlib import Path
from sklearn.metrics import confusion_matrix, classification_report

np.random.seed(42)
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['figure.dpi'] = 80

DATA_DIR = Path('.')
MODEL_DIR = Path('../models')
MODEL_DIR.mkdir(exist_ok=True)
print('Data dir:', DATA_DIR.resolve())
print('Model dir:', MODEL_DIR.resolve())
```

## 1. Load dữ liệu MNIST từ file idx

```python
def load_mnist_images(path):
    with open(path, 'rb') as f:
        magic, num, rows, cols = struct.unpack('>IIII', f.read(16))
        data = np.frombuffer(f.read(), dtype=np.uint8).reshape(num, rows, cols)
    return data

def load_mnist_labels(path):
    with open(path, 'rb') as f:
        magic, num = struct.unpack('>II', f.read(8))
        data = np.frombuffer(f.read(), dtype=np.uint8)
    return data

X_train_full = load_mnist_images(DATA_DIR / 'train-images.idx3-ubyte')
y_train_full = load_mnist_labels(DATA_DIR / 'train-labels.idx1-ubyte')
X_test = load_mnist_images(DATA_DIR / 't10k-images.idx3-ubyte')
y_test = load_mnist_labels(DATA_DIR / 't10k-labels.idx1-ubyte')

print('Train:', X_train_full.shape, y_train_full.shape)
print('Test :', X_test.shape, y_test.shape)
```

### Giảm kích thước để train nhanh (subsample)

```python
# Dùng 12k mẫu train để train nhanh (numpy thuần rất chậm)
N_TRAIN = 12000
N_VAL = 2000

idx = np.random.permutation(len(X_train_full))
X_train = X_train_full[idx[:N_TRAIN]]
y_train = y_train_full[idx[:N_TRAIN]]
X_val = X_train_full[idx[N_TRAIN:N_TRAIN+N_VAL]]
y_val = y_train_full[idx[N_TRAIN:N_TRAIN+N_VAL]]

# Normalize
X_train = (X_train.astype(np.float32) / 255.0)[:, None, :, :]  # NCHW
X_val   = (X_val.astype(np.float32)   / 255.0)[:, None, :, :]
X_test  = (X_test.astype(np.float32)  / 255.0)[:, None, :, :]

# Downsample 28x28 -> 14x14 để numpy train nhanh hơn
def downsample(X):
    return X[:, :, ::2, ::2]

X_train = downsample(X_train)
X_val   = downsample(X_val)
X_test_ds = downsample(X_test)
print('After downsample:', X_train.shape, X_val.shape, X_test_ds.shape)
```

## 2. EDA - Khám phá dữ liệu

```python
fig, axes = plt.subplots(3, 8, figsize=(14, 6))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train[i, 0], cmap='gray')
    ax.set_title(f'Label: {y_train[i]}', fontsize=9)
    ax.axis('off')
plt.suptitle('Mẫu ảnh MNIST (đã downsample 14x14)', fontsize=13)
plt.tight_layout()
plt.show()
```

## 3. Các layer cơ bản (NumPy)

```python
def im2col(X, k, s, p):
    N, C, H, W = X.shape
    H_out = (H + 2*p - k)//s + 1
    W_out = (W + 2*p - k)//s + 1
    Xp = np.pad(X, ((0,0),(0,0),(p,p),(p,p)))
    cols = np.zeros((N, C, k, k, H_out, W_out), dtype=np.float32)
    for i in range(k):
        for j in range(k):
            cols[:,:,i,j,:,:] = Xp[:,:,i:i+H_out*s:s, j:j+W_out*s:s]
    return cols.transpose(0,4,5,1,2,3).reshape(N*H_out*W_out, -1), H_out, W_out

class Conv2D:
    def __init__(self, in_c, out_c, k=3, s=1, p=1):
        self.W = np.random.randn(out_c, in_c, k, k).astype(np.float32) * np.sqrt(2.0/(in_c*k*k))
        self.b = np.zeros(out_c, dtype=np.float32)
        self.k, self.s, self.p = k, s, p
        self.in_c, self.out_c = in_c, out_c
    def forward(self, X):
        self.X_shape = X.shape
        cols, H_out, W_out = im2col(X, self.k, self.s, self.p)
        self.cols = cols
        W_flat = self.W.reshape(self.out_c, -1)
        out = cols @ W_flat.T + self.b  # (N*H_out*W_out, out_c)
        return out.reshape(self.X_shape[0], H_out, W_out, self.out_c).transpose(0,3,1,2)
    def backward(self, dout, lr):
        N, _, H_out, W_out = dout.shape
        dout_flat = dout.transpose(0,2,3,1).reshape(-1, self.out_c)
        dW = (dout_flat.T @ self.cols).reshape(self.W.shape)
        db = dout_flat.sum(axis=0)
        dcols = dout_flat @ self.W.reshape(self.out_c, -1)
        dcols = dcols.reshape(N, H_out, W_out, self.in_c, self.k, self.k).transpose(0,3,4,5,1,2)
        dX = np.zeros(self.X_shape, dtype=np.float32)
        # simplified col2im via scatter (slow but ok for 14x14)
        H = self.X_shape[2]; W = self.X_shape[3]
        dXp = np.zeros((N, self.in_c, H+2*self.p, W+2*self.p), dtype=np.float32)
        for i in range(self.k):
            for j in range(self.k):
                dXp[:,:,i:i+H_out*self.s:self.s, j:j+W_out*self.s:self.s] += dcols[:,:,:,:,i,j].transpose(0,1,3,2)
        if self.p > 0:
            dX = dXp[:,:,self.p:-self.p, self.p:-self.p]
        else:
            dX = dXp
        self.W -= lr * dW
        self.b -= lr * db
        return dX

def relu(x): return np.maximum(0, x)
def relu_grad(x): return (x > 0).astype(np.float32)

class MaxPool2D:
    def __init__(self, k=2, s=2):
        self.k, self.s = k, s
    def forward(self, X):
        N, C, H, W = X.shape
        k, s = self.k, self.s
        H_out = (H-k)//s + 1; W_out = (W-k)//s + 1
        self.X = X; self.H_out = H_out; self.W_out = W_out
        out = np.zeros((N, C, H_out, W_out), dtype=np.float32)
        for i in range(H_out):
            for j in range(W_out):
                patch = X[:,:,i*s:i*s+k, j*s:j*s+k]
                out[:,:,i,j] = patch.max(axis=(2,3))
        return out
    def backward(self, dout):
        dX = np.zeros_like(self.X)
        k, s = self.k, self.s
        for i in range(self.H_out):
            for j in range(self.W_out):
                patch = self.X[:,:,i*s:i*s+k, j*s:j*s+k]
                mask = (patch == patch.max(axis=(2,3), keepdims=True))
                dX[:,:,i*s:i*s+k, j*s:j*s+k] += mask * dout[:,:,i:i+1,j:j+1]
        return dX

class Flatten:
    def forward(self, X): self.shape = X.shape; return X.reshape(X.shape[0], -1)
    def backward(self, d): return d.reshape(self.shape)

class Dense:
    def __init__(self, in_f, out_f):
        self.W = np.random.randn(in_f, out_f).astype(np.float32) * np.sqrt(2.0/in_f)
        self.b = np.zeros(out_f, dtype=np.float32)
    def forward(self, X):
        self.X = X
        return X @ self.W + self.b
    def backward(self, dout, lr):
        dW = self.X.T @ dout
        db = dout.sum(axis=0)
        dX = dout @ self.W.T
        self.W -= lr * dW; self.b -= lr * db
        return dX

def softmax(x):
    x = x - x.max(axis=1, keepdims=True)
    e = np.exp(x)
    return e / e.sum(axis=1, keepdims=True)
```

## 4. Mô hình CNN (classifier)

```python
class CNN:
    def __init__(self):
        self.conv1 = Conv2D(1, 8, k=3, s=1, p=1)   # -> 8 x 14 x 14
        self.pool1 = MaxPool2D(2, 2)                # -> 8 x 7 x 7
        self.conv2 = Conv2D(8, 16, k=3, s=1, p=1)  # -> 16 x 7 x 7
        self.pool2 = MaxPool2D(2, 2)                # -> 16 x 3 x 3
        self.flat = Flatten()
        self.fc1 = Dense(16*3*3, 64)
        self.fc2 = Dense(64, 10)
    def forward(self, X):
        x = self.conv1.forward(X); self.a1 = x; x = relu(x)
        x = self.pool1.forward(x)
        x = self.conv2.forward(x); self.a2 = x; x = relu(x)
        x = self.pool2.forward(x)
        x = self.flat.forward(x)
        x = self.fc1.forward(x); self.a3 = x; x = relu(x)
        x = self.fc2.forward(x)
        return softmax(x)
    def backward(self, probs, y, lr):
        N = probs.shape[0]
        d = probs.copy(); d[np.arange(N), y] -= 1; d /= N
        d = self.fc2.backward(d, lr)
        d = d * relu_grad(self.a3)
        d = self.fc1.backward(d, lr)
        d = self.flat.backward(d)
        d = self.pool2.backward(d)
        d = d * relu_grad(self.a2)
        d = self.conv2.backward(d, lr)
        d = self.pool1.backward(d)
        d = d * relu_grad(self.a1)
        self.conv1.backward(d, lr)
```

## 5. Training

```python
def accuracy(model, X, y, bs=256):
    correct = 0
    for i in range(0, len(X), bs):
        p = model.forward(X[i:i+bs])
        correct += (p.argmax(1) == y[i:i+bs]).sum()
    return correct / len(X)

def cross_entropy(probs, y):
    eps = 1e-9
    return -np.log(probs[np.arange(len(y)), y] + eps).mean()

model = CNN()
EPOCHS = 6
BATCH = 64
LR = 0.01

history = {'loss': [], 'val_loss': [], 'acc': [], 'val_acc': []}
start = time.time()

for ep in range(EPOCHS):
    idx = np.random.permutation(len(X_train))
    Xs, ys = X_train[idx], y_train[idx]
    ep_loss = 0; n_batch = 0
    for i in range(0, len(Xs), BATCH):
        xb, yb = Xs[i:i+BATCH], ys[i:i+BATCH]
        p = model.forward(xb)
        ep_loss += cross_entropy(p, yb)
        n_batch += 1
        model.backward(p, yb, LR)
    tr_loss = ep_loss / n_batch
    tr_acc = accuracy(model, X_train, y_train)
    # val
    val_probs = []
    for i in range(0, len(X_val), 256):
        val_probs.append(model.forward(X_val[i:i+256]))
    val_probs = np.concatenate(val_probs)
    val_loss = cross_entropy(val_probs, y_val)
    val_acc = (val_probs.argmax(1) == y_val).mean()
    history['loss'].append(tr_loss); history['val_loss'].append(val_loss)
    history['acc'].append(tr_acc); history['val_acc'].append(val_acc)
    print(f'Epoch {ep+1}/{EPOCHS} - loss {tr_loss:.4f} - acc {tr_acc:.4f} - val_loss {val_loss:.4f} - val_acc {val_acc:.4f}')

print(f'Total time: {time.time()-start:.1f}s')
```

## 6. Learning Curves

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
axes[0].plot(history['loss'], 'o-', label='Train')
axes[0].plot(history['val_loss'], 's-', label='Val')
axes[0].set_title('Loss Curve'); axes[0].set_xlabel('Epoch'); axes[0].set_ylabel('Loss')
axes[0].legend(); axes[0].grid(alpha=0.3)
axes[1].plot(history['acc'], 'o-', label='Train')
axes[1].plot(history['val_acc'], 's-', label='Val')
axes[1].set_title('Accuracy Curve'); axes[1].set_xlabel('Epoch'); axes[1].set_ylabel('Accuracy')
axes[1].legend(); axes[1].grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 7. Đánh giá trên Test set

```python
test_probs = []
for i in range(0, len(X_test_ds), 256):
    test_probs.append(model.forward(X_test_ds[i:i+256]))
test_probs = np.concatenate(test_probs)
y_pred = test_probs.argmax(1)
test_acc = (y_pred == y_test).mean()
print(f'Test Accuracy: {test_acc:.4f}')
```

## 8. Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(9, 7))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=True)
plt.title('Confusion Matrix - Test set')
plt.xlabel('Predicted'); plt.ylabel('True')
plt.show()

cm_norm = cm.astype(float) / cm.sum(axis=1, keepdims=True)
plt.figure(figsize=(9, 7))
sns.heatmap(cm_norm, annot=True, fmt='.2f', cmap='Greens', cbar=True)
plt.title('Confusion Matrix (Normalized)')
plt.xlabel('Predicted'); plt.ylabel('True')
plt.show()
```

## 9. Classification Report

```python
print(classification_report(y_test, y_pred, digits=4))
```

## 10. Xem dự đoán sai

```python
wrong = np.where(y_pred != y_test)[0]
print(f'Số mẫu dự đoán sai: {len(wrong)}/{len(y_test)}')
fig, axes = plt.subplots(2, 8, figsize=(14, 4))
for i, ax in enumerate(axes.flat):
    k = wrong[i]
    ax.imshow(X_test_ds[k, 0], cmap='gray')
    ax.set_title(f'T:{y_test[k]} / P:{y_pred[k]}', fontsize=9, color='red')
    ax.axis('off')
plt.suptitle('Một số mẫu dự đoán sai', fontsize=12)
plt.tight_layout(); plt.show()
```

## 11. Lưu best model

```python
with open(MODEL_DIR / 'mnist_cnn_numpy.pkl', 'wb') as f:
    pickle.dump({
        'conv1_W': model.conv1.W, 'conv1_b': model.conv1.b,
        'conv2_W': model.conv2.W, 'conv2_b': model.conv2.b,
        'fc1_W': model.fc1.W, 'fc1_b': model.fc1.b,
        'fc2_W': model.fc2.W, 'fc2_b': model.fc2.b,
        'test_acc': test_acc, 'history': history
    }, f)
print('Saved:', (MODEL_DIR / 'mnist_cnn_numpy.pkl').resolve())
```
