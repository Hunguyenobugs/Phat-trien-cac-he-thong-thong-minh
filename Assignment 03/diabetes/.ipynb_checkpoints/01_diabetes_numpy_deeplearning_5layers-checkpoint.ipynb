# Deep Learning from Scratch — Diabetes Prediction
### NumPy implementation, 5 trainable layers, and comparison with 3 classical ML models

**Nguồn lý thuyết:** notebook bám theo hai tài liệu Assignment/Lecture 03 được cung cấp. Tài liệu nhấn mạnh: neural network là hàm tham số hóa; learned representation; linear transformation + nonlinearity; ReLU; forward propagation; loss; backpropagation/chain rule; gradient descent; mini-batch; training/validation curves; underfitting/overfitting; và Accuracy, Precision, Recall, F1, Confusion Matrix. 

**Lưu ý về thư viện:** phần học máy và deep learning đều tự cài đặt bằng NumPy. `matplotlib` chỉ dùng để trực quan hóa; không dùng PyTorch, TensorFlow, Keras hoặc scikit-learn.

```python
import numpy as np
import matplotlib.pyplot as plt
import csv, time, math
from pathlib import Path
np.set_printoptions(suppress=True, precision=4)
SEED = 42
np.random.seed(SEED)
print("NumPy:", np.__version__)
```

```python
# =========================
# NUMPY-ONLY ML/DL UTILITIES
# =========================

def read_csv_rows(path):
    path = Path(path)
    with path.open("r", encoding="utf-8-sig", newline="") as f:
        return list(csv.DictReader(f))

def is_missing(v):
    if v is None: return True
    s = str(v).strip()
    return s == "" or s.lower() in {"na","nan","null","none","?","missing"}

def try_float(v):
    if is_missing(v): return np.nan
    try: return float(str(v).strip())
    except: return np.nan

def column_types(rows):
    cols = list(rows[0].keys())
    numeric, categorical = [], []
    for c in cols:
        vals = [r[c] for r in rows[:min(500, len(rows))] if not is_missing(r[c])]
        ok = sum(try_float(v) == try_float(v) for v in vals)
        if vals and ok / len(vals) >= 0.95:
            numeric.append(c)
        else:
            categorical.append(c)
    return numeric, categorical

def encode_table(rows, target_col, drop_cols=None):
    drop_cols = set(drop_cols or [])
    cols = [c for c in rows[0].keys() if c != target_col and c not in drop_cols]
    numeric, categorical = column_types(rows)
    numeric = [c for c in numeric if c in cols]
    categorical = [c for c in categorical if c in cols]

    X_parts, feature_names = [], []
    for c in numeric:
        x = np.array([try_float(r[c]) for r in rows], dtype=float)
        med = np.nanmedian(x) if np.any(np.isfinite(x)) else 0.0
        x[~np.isfinite(x)] = med
        X_parts.append(x[:,None])
        feature_names.append(c)

    for c in categorical:
        vals = [str(r[c]).strip() if not is_missing(r[c]) else "__MISSING__" for r in rows]
        cats = sorted(set(vals))
        # avoid a huge one-hot explosion for very high-cardinality columns
        if len(cats) > 50:
            counts = {v: vals.count(v) for v in cats}
            keep = sorted(cats, key=lambda z: counts[z], reverse=True)[:49]
            vals = [v if v in keep else "__OTHER__" for v in vals]
            cats = sorted(set(vals))
        mat = np.zeros((len(rows), len(cats)))
        for j, cat in enumerate(cats):
            mat[:,j] = np.array([v == cat for v in vals], dtype=float)
            feature_names.append(f"{c}={cat}")
        X_parts.append(mat)

    X = np.hstack(X_parts) if X_parts else np.zeros((len(rows),0))
    y_raw = [r[target_col] for r in rows]
    return X, y_raw, feature_names

def train_val_test_split(X, y, task="classification", seed=42):
    rng = np.random.default_rng(seed)
    idx = rng.permutation(len(X))
    if task == "classification":
        # simple stratified 70/15/15 split implemented with NumPy
        train, val, test = [], [], []
        for cls in np.unique(y):
            ids = idx[y[idx] == cls]
            n=len(ids); a=int(.70*n); b=int(.85*n)
            train.extend(ids[:a]); val.extend(ids[a:b]); test.extend(ids[b:])
        train, val, test = map(np.array, (train,val,test))
        rng.shuffle(train); rng.shuffle(val); rng.shuffle(test)
    else:
        n=len(idx); a=int(.70*n); b=int(.85*n)
        train,val,test=idx[:a],idx[a:b],idx[b:]
    return X[train], X[val], X[test], y[train], y[val], y[test]

def standardize_fit(X):
    mu = X.mean(axis=0)
    sd = X.std(axis=0)
    sd[sd < 1e-8] = 1.0
    return mu, sd

def standardize_apply(X, mu, sd):
    return (X-mu)/sd

def accuracy(y,p): return float(np.mean(y==p))
def precision_recall_f1(y,p):
    tp=np.sum((y==1)&(p==1)); fp=np.sum((y==0)&(p==1))
    fn=np.sum((y==1)&(p==0))
    prec=tp/(tp+fp+1e-12); rec=tp/(tp+fn+1e-12)
    f1=2*prec*rec/(prec+rec+1e-12)
    return float(prec),float(rec),float(f1)
def confusion(y,p):
    return np.array([[np.sum((y==0)&(p==0)),np.sum((y==0)&(p==1))],
                     [np.sum((y==1)&(p==0)),np.sum((y==1)&(p==1))]])
def binary_auc(y, score):
    order=np.argsort(-score); ys=y[order]
    pos=np.sum(y==1); neg=np.sum(y==0)
    if pos==0 or neg==0: return np.nan
    tps=np.cumsum(ys==1); fps=np.cumsum(ys==0)
    tpr=tps/pos; fpr=fps/neg
    return float(np.trapezoid(np.r_[0,fpr,1], np.r_[0,tpr,1]))

def plot_confusion(cm,title):
    plt.figure(figsize=(4,3))
    plt.imshow(cm, interpolation="nearest")
    plt.title(title); plt.xlabel("Predicted"); plt.ylabel("Actual")
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j,i,str(cm[i,j]),ha="center",va="center")
    plt.colorbar(); plt.tight_layout(); plt.show()

def count_params(sizes):
    return int(sum(sizes[i]*sizes[i+1]+sizes[i+1] for i in range(len(sizes)-1)))
```

```python
# =========================
# THREE BASIC ML MODELS
# =========================

def sigmoid(z):
    z=np.clip(z,-50,50)
    return 1/(1+np.exp(-z))

def logistic_fit(X,y,lr=0.01,epochs=500,l2=0.0):
    w=np.zeros(X.shape[1]); b=0.0
    for _ in range(epochs):
        p=sigmoid(X@w+b)
        dw=X.T@(p-y)/len(X)+l2*w
        db=np.mean(p-y)
        w-=lr*dw; b-=lr*db
    return w,b

def logistic_predict(X,wb):
    w,b=wb; return sigmoid(X@w+b)

def knn_predict(Xtr,ytr,Xte,k=7):
    out=[]
    for x in Xte:
        d=np.sum((Xtr-x)**2,axis=1)
        ids=np.argpartition(d,min(k-1,len(d)-1))[:k]
        vals=ytr[ids]
        out.append(1 if np.mean(vals)>=0.5 else 0)
    return np.array(out)

def linear_fit(X,y,l2=0.0):
    A=X.T@X+l2*np.eye(X.shape[1])
    w=np.linalg.pinv(A)@X.T@y
    return w

def linear_predict(X,w): return X@w

def knn_reg_predict(Xtr,ytr,Xte,k=7):
    out=[]
    for x in Xte:
        d=np.sum((Xtr-x)**2,axis=1)
        ids=np.argpartition(d,min(k-1,len(d)-1))[:k]
        out.append(np.mean(ytr[ids]))
    return np.array(out)

def classification_ml_metrics(y,p,score=None):
    pr,re,f1=precision_recall_f1(y,p)
    return {"Accuracy":accuracy(y,p),"Precision":pr,"Recall":re,"F1":f1,
            "ROC-AUC":binary_auc(y,score if score is not None else p.astype(float))}

def regression_metrics(y,p):
    e=y-p
    mae=np.mean(np.abs(e)); rmse=np.sqrt(np.mean(e**2))
    ssr=np.sum(e**2); sst=np.sum((y-y.mean())**2)
    r2=1-ssr/(sst+1e-12)
    return {"MAE":float(mae),"RMSE":float(rmse),"R2":float(r2)}
```

```python
# ==========================================
# 5-TRAINABLE-LAYER DEEP NETWORK FROM SCRATCH
# ==========================================
# Architecture:
# d -> 64 -> 32 -> 16 -> 8 -> output
# Five trainable affine layers. ReLU is used between hidden layers.
# Binary classification: sigmoid + binary cross-entropy.
# Regression: linear output + MSE.

def relu(x): return np.maximum(0,x)
def relu_derivative(x): return (x>0).astype(float)

class NumpyDeepNet:
    def __init__(self,input_dim,output_type="binary",hidden=(64,32,16,8),seed=42):
        self.output_type=output_type
        sizes=[input_dim,*hidden,1]
        rng=np.random.default_rng(seed)
        self.W=[]; self.b=[]
        for a,b in zip(sizes[:-1],sizes[1:]):
            self.W.append(rng.standard_normal((a,b))*np.sqrt(2.0/a))
            self.b.append(np.zeros((1,b)))

    def forward(self,X,cache=True):
        A=X; caches=[]
        for i in range(len(self.W)-1):
            Z=A@self.W[i]+self.b[i]
            H=relu(Z)
            caches.append((A,Z,H))
            A=H
        Z=A@self.W[-1]+self.b[-1]
        if self.output_type=="binary": out=sigmoid(Z)
        else: out=Z
        if cache: return out,caches,Z,A
        return out

    def loss(self,y,p):
        y=y.reshape(-1,1)
        if self.output_type=="binary":
            p=np.clip(p,1e-8,1-1e-8)
            return float(-np.mean(y*np.log(p)+(1-y)*np.log(1-p)))
        return float(np.mean((p-y)**2)/2)

    def backward(self,y,p,caches,Z,last_A):
        y=y.reshape(-1,1); n=len(y)
        gradsW=[None]*len(self.W); gradsb=[None]*len(self.b)
        if self.output_type=="binary":
            dZ=(p-y)/n
        else:
            dZ=(p-y)/n
        gradsW[-1]=last_A.T@dZ
        gradsb[-1]=np.sum(dZ,axis=0,keepdims=True)
        dA=dZ@self.W[-1].T
        for i in range(len(self.W)-2,-1,-1):
            A,Zi,H=caches[i]
            dZ=dA*relu_derivative(Zi)
            gradsW[i]=A.T@dZ
            gradsb[i]=np.sum(dZ,axis=0,keepdims=True)
            if i>0: dA=dZ@self.W[i].T
        return gradsW,gradsb

    def update_sgd(self,gW,gb,lr):
        for i in range(len(self.W)):
            self.W[i]-=lr*gW[i]; self.b[i]-=lr*gb[i]

    def update_momentum(self,gW,gb,lr,beta,state):
        if state is None:
            state=([np.zeros_like(w) for w in self.W],
                   [np.zeros_like(b) for b in self.b])
        vw,vb=state
        for i in range(len(self.W)):
            vw[i]=beta*vw[i]+(1-beta)*gW[i]
            vb[i]=beta*vb[i]+(1-beta)*gb[i]
            self.W[i]-=lr*vw[i]; self.b[i]-=lr*vb[i]
        return state

    def update_adam(self,gW,gb,lr,beta1,beta2,eps,state,t):
        if state is None:
            state=([np.zeros_like(w) for w in self.W],
                   [np.zeros_like(b) for b in self.b],
                   [np.zeros_like(w) for w in self.W],
                   [np.zeros_like(b) for b in self.b])
        mw,mb,vw,vb=state
        for i in range(len(self.W)):
            mw[i]=beta1*mw[i]+(1-beta1)*gW[i]
            mb[i]=beta1*mb[i]+(1-beta1)*gb[i]
            vw[i]=beta2*vw[i]+(1-beta2)*(gW[i]**2)
            vb[i]=beta2*vb[i]+(1-beta2)*(gb[i]**2)
            mwc=mw[i]/(1-beta1**t); mbc=mb[i]/(1-beta1**t)
            vwc=vw[i]/(1-beta2**t); vbc=vb[i]/(1-beta2**t)
            self.W[i]-=lr*mwc/(np.sqrt(vwc)+eps)
            self.b[i]-=lr*mbc/(np.sqrt(vbc)+eps)
        return state

def batch_indices(n,batch_size,rng):
    ids=rng.permutation(n)
    for s in range(0,n,batch_size):
        yield ids[s:s+batch_size]

def nn_predict(model,X):
    p=model.forward(X,cache=False)
    if model.output_type=="binary": return p.reshape(-1), (p.reshape(-1)>=0.5).astype(int)
    return p.reshape(-1), p.reshape(-1)

def train_nn(model,Xtr,ytr,Xv,yv,epochs=300,batch_size=32,lr=0.001,
             optimizer="adam",patience=50,seed=42):
    rng=np.random.default_rng(seed); hist={"train_loss":[],"val_loss":[],"train_metric":[],"val_metric":[]}
    state=None; best=None; best_val=np.inf; wait=0; t0=time.time()
    for ep in range(1,epochs+1):
        for ids in batch_indices(len(Xtr),batch_size,rng):
            p,c,Z,A=model.forward(Xtr[ids],cache=True)
            gW,gb=model.backward(ytr[ids],p,c,Z,A)
            if optimizer=="sgd":
                model.update_sgd(gW,gb,lr)
            elif optimizer=="momentum":
                state=model.update_momentum(gW,gb,lr,0.9,state)
            else:
                state=model.update_adam(gW,gb,lr,0.9,0.999,1e-8,state,ep)
        ptr=model.forward(Xtr,cache=False); pv=model.forward(Xv,cache=False)
        tl=model.loss(ytr,ptr); vl=model.loss(yv,pv)
        hist["train_loss"].append(tl); hist["val_loss"].append(vl)
        if model.output_type=="binary":
            hist["train_metric"].append(accuracy(ytr,(ptr.reshape(-1)>=.5).astype(int)))
            hist["val_metric"].append(accuracy(yv,(pv.reshape(-1)>=.5).astype(int)))
        else:
            hist["train_metric"].append(np.sqrt(np.mean((ptr.reshape(-1)-ytr)**2)))
            hist["val_metric"].append(np.sqrt(np.mean((pv.reshape(-1)-yv)**2)))
        if vl < best_val:
            best_val=vl; wait=0
            best=([w.copy() for w in model.W],[b.copy() for b in model.b])
        else:
            wait+=1
            if wait>=patience: break
    if best is not None:
        model.W,model.b=best
    return hist,time.time()-t0,ep

def plot_history(hist,metric_name="Accuracy"):
    fig=plt.figure(figsize=(12,4))
    ax=fig.add_subplot(1,2,1); ax.plot(hist["train_loss"],label="Train"); ax.plot(hist["val_loss"],label="Validation")
    ax.set_title("Loss curves"); ax.set_xlabel("Epoch"); ax.set_ylabel("Loss"); ax.legend(); ax.grid(alpha=.25)
    ax=fig.add_subplot(1,2,2); ax.plot(hist["train_metric"],label="Train"); ax.plot(hist["val_metric"],label="Validation")
    ax.set_title(metric_name+" curves"); ax.set_xlabel("Epoch"); ax.set_ylabel(metric_name); ax.legend(); ax.grid(alpha=.25)
    plt.tight_layout(); plt.show()

def plot_learned_representation(model,X,y):
    # Inspect h1 -> h2 -> ... and visualize first two dimensions.
    A=X
    reps=[]
    for i in range(len(model.W)-1):
        A=relu(A@model.W[i]+model.b[i]); reps.append(A.copy())
    for i,H in enumerate(reps):
        if H.shape[1]>=2:
            plt.figure(figsize=(5,4))
            plt.scatter(H[y==0,0],H[y==0,1],alpha=.35,label="Class 0")
            plt.scatter(H[y==1,0],H[y==1,1],alpha=.35,label="Class 1")
            plt.title(f"Learned representation H{i+1}: first 2 dims")
            plt.xlabel("dim 1"); plt.ylabel("dim 2"); plt.legend(); plt.grid(alpha=.2); plt.show()
```

## 1. Load và kiểm tra dữ liệu
Dataset diabetes hiện có là bản messy 100,000 dòng với 34 cột. Target được chọn là `diagnosed_diabetes`. Các cột định danh/phụ trợ như `patient_tracking_id`, `data_entry_notes` không được dùng. `diabetes_stage` bị loại để tránh target leakage.

Quy trình giữ đúng tinh thần tài liệu: **Raw data → Clean → Numerical representation → Model input**.

```python
DATA_PATH = "/mnt/data/final_messy_diabetes_dataset(1).csv"
rows = read_csv_rows(DATA_PATH)
print("Rows:", len(rows))
print("Columns:", len(rows[0]))
print("First record:")
for k,v in list(rows[0].items())[:12]:
    print(f"  {k}: {v}")
```

```python
cols=list(rows[0].keys())
missing=[]
for c in cols:
    n=sum(is_missing(r[c]) for r in rows)
    missing.append((c,n,n/len(rows)*100))
print("Missing-value summary:")
for x in sorted(missing,key=lambda z:-z[1])[:15]:
    print(f"{x[0]:30s} {x[1]:7d} ({x[2]:5.1f}%)")
print("\nTarget distribution:")
yraw=np.array([int(float(r["diagnosed_diabetes"])) for r in rows])
for c in np.unique(yraw):
    print(c, np.sum(yraw==c), f"{np.mean(yraw==c)*100:.2f}%")
```

```python
# Clean + encode specifically for this diabetes dataset.
DROP_COLS={"patient_tracking_id","data_entry_notes","diabetes_stage"}
X_raw,y_raw,feature_names=encode_table(rows,"diagnosed_diabetes",DROP_COLS)
y=np.array([int(float(v)) for v in y_raw])
print("Feature matrix:",X_raw.shape)
print("Target:",y.shape)
print("First 15 model features:",feature_names[:15])
```

## 3. Train / validation / test và chuẩn hóa

Tài liệu khuyến nghị tách độc lập train/validation/test; ở đây dùng **70% / 15% / 15%** và giữ validation/test không tham gia fit preprocessing. Standardization dùng mean/std của training set.

```python
Xtr,Xv,Xte,ytr,yv,yte=train_val_test_split(X_raw,y,"classification",SEED)
mu,sd=standardize_fit(Xtr)
Xtr=standardize_apply(Xtr,mu,sd); Xv=standardize_apply(Xv,mu,sd); Xte=standardize_apply(Xte,mu,sd)
print(Xtr.shape,Xv.shape,Xte.shape)
print("Class balance train/val/test:")
for name,yy in [("train",ytr),("val",yv),("test",yte)]:
    print(name, np.bincount(yy,minlength=2), np.mean(yy))
```

## 4. Ba mô hình Machine Learning cơ bản

Để đáp ứng yêu cầu so sánh, ba mô hình được tự cài đặt bằng NumPy:

1. **Logistic Regression** — mô hình tuyến tính.
2. **KNN** — dựa trên khoảng cách, không học representation sâu.
3. **Decision Stump** — cây quyết định 1 mức, đại diện cho mô hình rule-based đơn giản.

Sau đó Deep Learning sẽ được so sánh trên cùng test set.

```python
# Logistic Regression
t=time.time()
wb=logistic_fit(Xtr,ytr,lr=.03,epochs=700,l2=1e-4)
score_lr=logistic_predict(Xte,wb); pred_lr=(score_lr>=.5).astype(int)
r_lr=classification_ml_metrics(yte,pred_lr,score_lr)

# KNN
t2=time.time()
pred_knn=knn_predict(Xtr,ytr,Xte,k=11)
r_knn=classification_ml_metrics(yte,pred_knn)

# Decision stump: choose feature + threshold minimizing classification error on train
best=(1e9,None,None)
for j in range(Xtr.shape[1]):
    vals=np.quantile(Xtr[:,j],np.linspace(.05,.95,19))
    for th in vals:
        for direction in [1,-1]:
            p=np.where(direction*Xtr[:,j]>=direction*th,1,0)
            err=np.mean(p!=ytr)
            if err<best[0]: best=(err,j,th,direction)
_,j,th,direction=best
pred_stump=np.where(direction*Xte[:,j]>=direction*th,1,0)
r_stump=classification_ml_metrics(yte,pred_stump)

results_ml=[]
for name,r in [("Logistic Regression",r_lr),("KNN",r_knn),("Decision Stump",r_stump)]:
    results_ml.append([name,*[r[k] for k in ["Accuracy","Precision","Recall","F1","ROC-AUC"]]])
print("Model | Accuracy | Precision | Recall | F1 | ROC-AUC")
for row in results_ml: print(row)
```

```python
arr=np.array([[r[1],r[2],r[3],r[4],0 if np.isnan(r[5]) else r[5]] for r in results_ml])
plt.figure(figsize=(10,4))
x=np.arange(3); w=.16
metrics=["Accuracy","Precision","Recall","F1","ROC-AUC"]
for i,m in enumerate(metrics): plt.bar(x+i*w,arr[:,i],w,label=m)
plt.xticks(x+2*w,["Logistic","KNN","Stump"]); plt.ylim(0,1.05)
plt.title("Classical ML comparison"); plt.legend(); plt.grid(axis="y",alpha=.2); plt.tight_layout(); plt.show()
for name,p in [("Logistic Regression",pred_lr),("KNN",pred_knn),("Decision Stump",pred_stump)]:
    plot_confusion(confusion(yte,p),f"Confusion Matrix — {name}")
```

## 5. Deep Learning lý thuyết và kiến trúc 5 lớp

Tài liệu mô tả mạng như một composition của các hàm học được:

\[
\hat y=f_3(f_2(f_1(X)))
\]

với linear layer \(Z=XW+b\), ReLU ở hidden layers, sigmoid cho binary classification, loss, backpropagation và gradient descent.

Ở bài này, để đúng yêu cầu **5 lớp trainable**, ta dùng:

**d → 64 → 32 → 16 → 8 → 1**

Tức 5 affine/trainable layers. Đây là mở rộng sâu hơn từ kiến trúc 8→16→8→1 trong tài liệu; bản chất vẫn là **representation → deeper representation → prediction**.

```python
INPUT_DIM=Xtr.shape[1]
ARCH=[INPUT_DIM,64,32,16,8,1]
print("Architecture:",ARCH)
print("Trainable layers:",len(ARCH)-1)
print("Trainable parameters:",count_params(ARCH))
for i,(a,b) in enumerate(zip(ARCH[:-1],ARCH[1:]),1):
    print(f"Layer {i}: {a} -> {b}, params={a*b+b}")
```

## 6. Kiểm tra forward propagation và representation

## 7. Controlled experiments — epochs, batch size, learning rate, optimizer

Theo Assignment 03, cần có baseline/deeper/optimization experiment và report architecture, parameter count, configuration, training/validation curves, test metrics và interpretation. Ở đây ta mở rộng thành nhiều thí nghiệm có kiểm soát.

Optimizer được tự cài đặt bằng NumPy:
- SGD
- Momentum
- Adam

Điều này thay thế phần framework optimizer trong slide nhưng giữ nguyên chu trình **Forward → Loss → Backward → Update**.

```python
EXPERIMENTS=[
    {"name":"SGD lr=0.01 bs=32 ep=250","lr":.01,"batch":32,"epochs":250,"opt":"sgd"},
    {"name":"SGD lr=0.003 bs=32 ep=350","lr":.003,"batch":32,"epochs":350,"opt":"sgd"},
    {"name":"Momentum lr=0.003 bs=32 ep=350","lr":.003,"batch":32,"epochs":350,"opt":"momentum"},
    {"name":"Adam lr=0.001 bs=32 ep=350","lr":.001,"batch":32,"epochs":350,"opt":"adam"},
    {"name":"Adam lr=0.0005 bs=64 ep=500","lr":.0005,"batch":64,"epochs":500,"opt":"adam"},
    {"name":"Adam lr=0.001 bs=128 ep=500","lr":.001,"batch":128,"epochs":500,"opt":"adam"},
]
exp_results=[]; histories={}
for cfg in EXPERIMENTS:
    model=NumpyDeepNet(INPUT_DIM,"binary",hidden=(64,32,16,8),seed=SEED)
    hist,sec,used=train_nn(model,Xtr,ytr,Xv,yv,**{k:cfg[k] for k in ["epochs","batch_size","lr"]},
                           optimizer=cfg["opt"],patience=60,seed=SEED)
    prob,pred=nn_predict(model,Xte)
    met=classification_ml_metrics(yte,pred,prob)
    exp_results.append([cfg["name"],cfg["opt"],cfg["lr"],cfg["batch"],used,
                        met["Accuracy"],met["Precision"],met["Recall"],met["F1"],met["ROC-AUC"],used])
    histories[cfg["name"]]=(hist,model)
    print(cfg["name"],"-> F1",round(met["F1"],4),"Acc",round(met["Accuracy"],4),"time",round(sec,2),"s")
```

```python
print("Experiment comparison")
print("Name | Optimizer | LR | Batch | Accuracy | Precision | Recall | F1 | AUC")
for r in exp_results:
    print(r[:10])

vals=np.array([[r[5],r[6],r[7],r[8],0 if np.isnan(r[9]) else r[9]] for r in exp_results])
plt.figure(figsize=(12,5))
x=np.arange(len(exp_results)); w=.16
for i,m in enumerate(["Accuracy","Precision","Recall","F1","ROC-AUC"]):
    plt.bar(x+i*w,vals[:,i],w,label=m)
plt.xticks(x+2*w,[r[0] for r in exp_results],rotation=55,ha="right")
plt.ylim(0,1.05); plt.title("Deep Learning hyperparameter comparison")
plt.legend(); plt.grid(axis="y",alpha=.2); plt.tight_layout(); plt.show()
```

## 8. Chọn mô hình tối ưu theo validation, sau đó đánh giá test

Không chọn tham số chỉ vì training accuracy cao. Theo tài liệu, cần quan sát training/validation để phát hiện underfitting/overfitting và cuối cùng dùng held-out test để đánh giá generalization.

```python
best_i=int(np.argmax([r[8] for r in exp_results])) # best validation proxy chosen by test here only for demo
# For a strict experimental report, choose by validation F1 before looking at test.
# Re-run below to choose based on validation F1.
val_scores=[]
for cfg in EXPERIMENTS:
    hist,model=histories[cfg["name"]]
    val_scores.append(max(hist["val_metric"]))
best_name=EXPERIMENTS[int(np.argmax(val_scores))]["name"]
best_model=histories[best_name][1]
best_prob,best_pred=nn_predict(best_model,Xte)
best_metrics=classification_ml_metrics(yte,best_pred,best_prob)
print("Selected by best validation accuracy:",best_name)
print(best_metrics)
plot_confusion(confusion(yte,best_pred),"Best 5-layer Deep Learning")
plot_learned_representation(best_model,Xte,yte)
```

## 10. Kết luận mẫu

- **Representation:** ML cơ bản học trực tiếp từ feature vector đã được chuẩn bị; Deep Learning tạo các hidden representations qua nhiều phép biến đổi có tham số.
- **Nonlinearity:** ReLU giúp mạng có khả năng biểu diễn phi tuyến; nếu chỉ xếp các phép biến đổi tuyến tính, toàn bộ mạng vẫn tương đương một phép biến đổi tuyến tính.
- **Optimization:** learning rate, batch size và optimizer ảnh hưởng trực tiếp tới tốc độ hội tụ và chất lượng validation.
- **Generalization:** mô hình được chọn dựa trên validation behavior rồi mới báo cáo test metrics.
- **So sánh:** Deep Learning không mặc định tốt hơn ML truyền thống; lợi ích phải được chứng minh bằng metric, độ ổn định validation, số tham số và chi phí tính toán.
