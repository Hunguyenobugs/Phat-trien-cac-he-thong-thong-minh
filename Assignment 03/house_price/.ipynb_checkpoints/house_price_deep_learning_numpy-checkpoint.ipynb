## 2. Load, inspect, clean and engineer features

```python

import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
PATH="/mnt/data/house_buying_dec29th_2025(1).csv"
raw=np.genfromtxt(PATH,delimiter=",",names=True,dtype=None,encoding="utf-8",invalid_raise=False)
print("Shape:",raw.shape); print("Columns:",raw.dtype.names)
for n in raw.dtype.names: print(n,raw[n].dtype,"unique=",len(np.unique(raw[n])))
def s(x):
    z="" if x is None else str(x); return "" if z.lower()=="nan" else " ".join(z.strip().split())
def f(x):
    try:return float(x)
    except:return np.nan
area=np.array([f(x) for x in raw["area_m2"]]); bed=np.array([f(x) for x in raw["bedrooms"]]); bath=np.array([f(x) for x in raw["bathrooms"]])
flo=np.array([f(x) for x in raw["floors"]]); time=np.array([f(x) for x in raw["timeline_hours"]]); front=np.array([1 if str(x).lower() in ("true","1") else 0 for x in raw["frontage"]],float)
loc=np.array([s(x) for x in raw["location"]]); title=np.array([s(x) for x in raw["title"]]); y=np.array([f(x) for x in raw["price_million_vnd"]]).reshape(-1,1)
def oh(a):
    cats,inv=np.unique(a,return_inverse=True); M=np.zeros((len(a),len(cats))); M[np.arange(len(a)),inv]=1; return M
L=oh(loc)
X=np.hstack([np.column_stack([area,bed,bath,flo,time,front,np.array([len(z) for z in title]),np.array([len(z.split()) for z in title])]),L])
valid=np.isfinite(y[:,0])&(y[:,0]>0); X,y=X[valid],y[valid]; X[~np.isfinite(X)]=np.nan
keep=np.unique(np.column_stack([np.nan_to_num(X,nan=-999999),y]),axis=0,return_index=True)[1]; X,y=X[np.sort(keep)],y[np.sort(keep)]
idx=np.random.permutation(len(X)); a=int(.70*len(X)); b=int(.85*len(X)); tr,va,te=idx[:a],idx[a:b],idx[b:]
Xtr,Xva,Xte=X[tr],X[va],X[te]; ytr,yva,yte=y[tr],y[va],y[te]
med=np.nanmedian(Xtr,0); Xtr=np.where(np.isfinite(Xtr),Xtr,med); Xva=np.where(np.isfinite(Xva),Xva,med); Xte=np.where(np.isfinite(Xte),Xte,med)
mu=Xtr.mean(0); sd=Xtr.std(0)+1e-8; Xtr=(Xtr-mu)/sd; Xva=(Xva-mu)/sd; Xte=(Xte-mu)/sd
ym=ytr.mean(); ys=ytr.std()+1e-8; ytr_s=(ytr-ym)/ys; yva_s=(yva-ym)/ys; yte_s=(yte-ym)/ys
print("After cleaning:",X.shape,"split:",Xtr.shape,Xva.shape,Xte.shape,"target mean:",ym)
```

## 3. Three basic ML models — NumPy only

```python

def linreg(X,y,lr=.01,epochs=80,batch=512):
    w=np.zeros((X.shape[1],1)); b=np.zeros((1,1)); rng=np.random.default_rng(1)
    for _ in range(epochs):
        order=rng.permutation(len(X))
        for s in range(0,len(X),batch):
            ix=order[s:s+batch]; d=2*(X[ix]@w+b-y[ix])/len(ix); w-=lr*X[ix].T@d; b-=lr*d.sum(0,keepdims=True)
    return w,b
def ridge(X,y,lr=.01,epochs=80,batch=512,l2=.01):
    w=np.zeros((X.shape[1],1)); b=np.zeros((1,1)); rng=np.random.default_rng(2)
    for _ in range(epochs):
        order=rng.permutation(len(X))
        for s in range(0,len(X),batch):
            ix=order[s:s+batch]; d=2*(X[ix]@w+b-y[ix])/len(ix); w-=lr*(X[ix].T@d+l2*w); b-=lr*d.sum(0,keepdims=True)
    return w,b
def knnr(X,y,Q,k=7):
    out=[]
    for q in Q:
        d=np.sum((X-q)**2,1); ix=np.argpartition(d,k)[:k]; out.append(y[ix].mean())
    return np.array(out).reshape(-1,1)
wl,bl=linreg(Xtr,ytr_s); wr,br=ridge(Xtr,ytr_s)
p1=(Xte@wl+bl)*ys+ym; p2=(Xte@wr+br)*ys+ym
n=min(2000,len(Xte)); p3=knnr(Xtr,ytr_s,Xte[:n])*ys+ym
base=[("Linear Regression",reg_metrics(yte,p1)),("Ridge Regression",reg_metrics(yte,p2)),("KNN Regression",reg_metrics(yte[:n],p3))]
for x in base: print(x)
```

## 5. Five-layer Deep Learning — NumPy

```python

import numpy as np
import matplotlib.pyplot as plt

def relu(z): return np.maximum(z,0)
def relu_grad(z): return (z>0).astype(float)
def init(dims,seed=42):
    rng=np.random.default_rng(seed); P={}
    for i in range(len(dims)-1):
        P[f"W{i+1}"]=rng.standard_normal((dims[i],dims[i+1]))*np.sqrt(2/dims[i]); P[f"b{i+1}"]=np.zeros((1,dims[i+1]))
    return P
def forward(X,P):
    c={"X":X}; A=X; L=len(P)//2
    for i in range(1,L):
        z=A@P[f"W{i}"]+P[f"b{i}"]; A=relu(z); c[f"Z{i}"],c[f"H{i}"]=z,A
    z=A@P[f"W{L}"]+P[f"b{L}"]; c[f"Z{L}"]=z; return z,c
def mse(y,p): return np.mean((p-y)**2)
def backward(y,c,P):
    L=len(P)//2; n=len(y); g={}; dz=2*(c[f"Z{L}"]-y)/n
    for i in range(L,0,-1):
        ap=c["X"] if i==1 else c[f"H{i-1}"]; g[f"W{i}"]=ap.T@dz; g[f"b{i}"]=dz.sum(0,keepdims=True)
        if i>1: dz=(dz@P[f"W{i}"].T)*relu_grad(c[f"Z{i-1}"])
    return g
def update(P,g,state,opt,lr,t):
    if opt=="sgd":
        for k in P:P[k]-=lr*g[k]
    elif opt=="momentum":
        for k in P: state[k]=.9*state[k]+.1*g[k]; P[k]-=lr*state[k]
    else:
        for k in P:
            state["m"][k]=.9*state["m"][k]+.1*g[k]; state["v"][k]=.999*state["v"][k]+.001*g[k]**2
            mh=state["m"][k]/(1-.9**t); vh=state["v"][k]/(1-.999**t); P[k]-=lr*mh/(np.sqrt(vh)+1e-8)
def train(X,y,Xv,yv,dims,epochs,batch,lr,opt,seed=42):
    P=init(dims,seed); state={} if opt=="sgd" else ({k:np.zeros_like(v) for k,v in P.items()} if opt=="momentum" else {"m":{k:np.zeros_like(v) for k,v in P.items()},"v":{k:np.zeros_like(v) for k,v in P.items()}})
    rng=np.random.default_rng(seed); H={"loss":[],"val":[]}
    for ep in range(1,epochs+1):
        order=rng.permutation(len(X))
        for s in range(0,len(X),batch):
            ix=order[s:s+batch]; p,c=forward(X[ix],P); g=backward(y[ix],c,P); update(P,g,state,opt,lr,ep)
        H["loss"].append(mse(y,forward(X,P)[0])); H["val"].append(mse(yv,forward(Xv,P)[0]))
    return P,H
def reg_metrics(y,p):
    e=p-y; mae=np.mean(np.abs(e)); rmse=np.sqrt(np.mean(e**2)); r2=1-np.sum(e**2)/(np.sum((y-y.mean())**2)+1e-8)
    return {"MAE":mae,"RMSE":rmse,"R2":r2}
```

```python

DIMS=[Xtr.shape[1],64,32,16,8,1]; print("5 trainable layers:",DIMS)
P,H=train(Xtr,ytr_s,Xva,yva_s,DIMS,160,512,.001,"adam"); pred=forward(Xte,P)[0]*ys+ym; print(reg_metrics(yte,pred))
plt.figure(figsize=(9,5)); plt.plot(H["loss"],label="train MSE"); plt.plot(H["val"],label="validation MSE"); plt.xlabel("Epoch"); plt.ylabel("MSE (standardized target)"); plt.title("5-layer DL loss curves"); plt.legend(); plt.grid(alpha=.25); plt.show()
```

## 6. Hyperparameter experiments

```python

configs=[("Adam",.001,256,120),("Adam",.001,512,160),("Adam",.0003,512,160),("SGD",.01,512,160),("SGD",.001,512,160),("Momentum",.01,512,160),("Adam",.003,512,120),("Adam",.001,1024,160)]
results=[]; curves=[]
for opt,lr,batch,ep in configs:
    q,h=train(Xtr,ytr_s,Xva,yva_s,DIMS,ep,batch,lr,opt); p=forward(Xte,q)[0]*ys+ym; m=reg_metrics(yte,p)
    results.append([opt,lr,batch,ep,m["MAE"],m["RMSE"],m["R2"],h["val"][-1]]); curves.append((opt+" lr="+str(lr)+" b="+str(batch),h))
print("opt lr batch epochs MAE RMSE R2 val_MSE")
for r in results: print(r)
```

```python

labels=[f"{r[0]} lr={r[1]} b={r[2]}" for r in results]
plt.figure(figsize=(12,5)); plt.bar(range(len(labels)),[r[6] for r in results]); plt.xticks(range(len(labels)),labels,rotation=70); plt.ylabel("R²"); plt.title("Hyperparameter comparison — R²"); plt.tight_layout(); plt.show()
plt.figure(figsize=(12,5)); plt.bar(range(len(labels)),[r[5] for r in results]); plt.xticks(range(len(labels)),labels,rotation=70); plt.ylabel("RMSE (million VND)"); plt.title("Hyperparameter comparison — RMSE"); plt.tight_layout(); plt.show()
plt.figure(figsize=(10,5))
for lab,h in curves[:6]: plt.plot(h["val"],label=lab)
plt.xlabel("Epoch"); plt.ylabel("Validation MSE"); plt.title("Validation curves"); plt.legend(fontsize=8); plt.grid(alpha=.25); plt.show()
```

## 7. Best model and ML vs DL

```python

bi=max(range(len(results)),key=lambda i:(results[i][6],-results[i][5])); opt,lr,batch,ep=configs[bi]
best,hh=train(Xtr,ytr_s,Xva,yva_s,DIMS,ep,batch,lr,opt); bestpred=forward(Xte,best)[0]*ys+ym; bm=reg_metrics(yte,bestpred)
print("BEST:",configs[bi]); print("DL:",bm)
allm=base+[("5-layer Deep Learning",bm)]
for n,m in allm: print(n,m)
plt.figure(figsize=(11,6)); plt.bar([n for n,_ in allm],[m["R2"] for _,m in allm]); plt.ylabel("R²"); plt.title("ML vs 5-layer DL — R²"); plt.xticks(rotation=15); plt.grid(axis="y",alpha=.25); plt.show()
```
