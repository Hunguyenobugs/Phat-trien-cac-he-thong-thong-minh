## 4. Làm sạch dữ liệu và loại bỏ target leakage

### 1) Code thực tế



### 2) Output thực tế



### 3) Giải thích code

Cell thực hiện làm sạch chính. Thứ nhất, thay `?` bằng `NaN`. Thứ hai, loại `patient_tracking_id` và `data_entry_notes`. Thứ ba, loại `diabetes_stage` vì đây là target leakage: bảng chéo cho thấy `Type 2` gần như xác định trực tiếp target 1 và `No Diabetes` xác định target 0.

### 4) Phân tích output

Output chứng minh `diabetes_stage` là leakage: `No Diabetes` chỉ đi với target 0, `Type 2` chỉ đi với target 1. Vì vậy loại biến này là cần thiết để tránh mô hình học một quy tắc gần như trực tiếp từ target.
