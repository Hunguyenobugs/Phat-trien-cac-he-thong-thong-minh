# E-commerce — User Interest Prediction with Deep Learning from Scratch

## Objective

This notebook predicts **User Interest** in a clothing product.

The dataset does not contain a column literally named `User Interest`. Therefore, following the available target in the dataset, we define:

- **Target:** `Recommended IND`
- `1` → the customer recommends the product → **Interested / positive interest signal**
- `0` → the customer does not recommend the product → **Not interested / negative interest signal**

This is a **binary classification** problem.

> Important: `Rating`, `Positive Feedback Count`, and other post-interaction signals are **not used as predictive inputs**, because they can leak information about the user's already-observed reaction.

## Course requirement

The Deep Learning model is implemented manually with **NumPy**:
- no PyTorch
- no TensorFlow
- no Keras
- no scikit-learn for the neural network

The notebook follows the theory in the course PDF:
**function composition → representation learning → forward propagation → loss → backpropagation → gradient descent → evaluation**.

The requested Deep Learning architecture has **5 trainable layers**:
`Input → 64 → 32 → 16 → 8 → 1`.


## Problem formulation

Let `X` contain information available for predicting interest:

- User: `Age`
- Product: `Clothing ID`
- Product hierarchy: `Division Name`, `Department Name`, `Class Name`
- Review text: `Title`, `Review Text`

The target is:

`y = Recommended IND ∈ {0, 1}`

The model learns:

`ŷ = f5(f4(f3(f2(f1(X)))))`

and the final sigmoid output is interpreted as:

`P(User Interest = 1 | X)`

A threshold of 0.5 is used for the default class prediction, while additional thresholds are tested later.


```python
import re
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
RANDOM_SEED = 42

DATA_PATH = "/mnt/data/Womens Clothing E-Commerce Reviews(1).csv"
```

```python
# Load CSV with NumPy only.
# The CSV is quoted, so genfromtxt is used for robust parsing.
raw = np.genfromtxt(
    DATA_PATH,
    delimiter=",",
    names=True,
    dtype=None,
    encoding="utf-8",
    missing_values="",
    filling_values=""
)

print("Columns:")
print(raw.dtype.names)
print("Rows:", len(raw))
```

## Target and leakage control

For **User Interest Prediction**, the target is `Recommended IND`.

The following fields are deliberately excluded from `X`:
- `Recommended IND` — target itself
- `Rating` — direct post-interaction satisfaction signal
- `Positive Feedback Count` — post-review engagement signal

This prevents the model from simply learning from information that is already a consequence of the user's interaction.


```python
# Inspect the target distribution
target = np.asarray(raw["Recommended_IND"], dtype=np.float64)

# Defensive conversion for the CSV field name produced by NumPy.
target = np.nan_to_num(target, nan=0.0)
target = (target >= 0.5).astype(np.int64)

classes, counts = np.unique(target, return_counts=True)
print("User Interest target distribution:")
for c, n in zip(classes, counts):
    print(f"Class {c}: {n} ({n/len(target)*100:.2f}%)")

plt.figure(figsize=(6,4))
plt.bar(["Not interested (0)", "Interested (1)"],
        [counts[0] if len(counts)>0 else 0,
         counts[1] if len(counts)>1 else 0])
plt.title("User Interest Target Distribution")
plt.ylabel("Number of reviews")
plt.show()
```

## NumPy feature engineering for User Interest

Text is represented using a lightweight **bag-of-words / token-frequency representation implemented directly with NumPy**.

Categorical fields are encoded with one-hot vectors implemented directly with NumPy.

No pandas, sklearn, TensorFlow, Keras, or PyTorch is required.


```python
def get_str(field, i):
    v = raw[field][i]
    if isinstance(v, bytes):
        v = v.decode("utf-8", errors="ignore")
    if v is None:
        return ""
    return str(v).strip()

# Build a compact text vocabulary from Title + Review Text.
# Keep the most frequent words to control dimensionality.
texts = []
for i in range(len(raw)):
    title = get_str("Title", i)
    review = get_str("Review_Text", i)
    txt = (title + " " + review).lower()
    tokens = re.findall(r"[a-z]+", txt)
    texts.append(tokens)

word_counts = {}
for tokens in texts:
    for w in set(tokens):
        word_counts[w] = word_counts.get(w, 0) + 1

MAX_VOCAB = 150
vocab = sorted(word_counts, key=word_counts.get, reverse=True)[:MAX_VOCAB]
word_to_idx = {w:i for i,w in enumerate(vocab)}

# Binary bag-of-words features.
X_text = np.zeros((len(raw), len(vocab)), dtype=np.float64)
for i,tokens in enumerate(texts):
    for w in set(tokens):
        j = word_to_idx.get(w)
        if j is not None:
            X_text[i,j] = 1.0

print("Text feature shape:", X_text.shape)
```

```python
def encode_categorical(field):
    values = np.array([get_str(field, i) for i in range(len(raw))], dtype=object)
    unique = sorted(set(values.tolist()))
    mapping = {v:i for i,v in enumerate(unique)}
    out = np.zeros((len(values), len(unique)), dtype=np.float64)
    for i,v in enumerate(values):
        out[i, mapping[v]] = 1.0
    return out, unique

# Numeric user/product information.
age = np.array([float(raw["Age"][i]) for i in range(len(raw))], dtype=np.float64)
clothing_id = np.array([float(raw["Clothing_ID"][i]) for i in range(len(raw))], dtype=np.float64)

# Robust missing-value handling.
age = np.nan_to_num(age, nan=np.nanmedian(age))
clothing_id = np.nan_to_num(clothing_id, nan=np.nanmedian(clothing_id))

X_num = np.column_stack([age, clothing_id])

X_div, div_values = encode_categorical("Division_Name")
X_dept, dept_values = encode_categorical("Department_Name")
X_class, class_values = encode_categorical("Class_Name")

X = np.hstack([X_num, X_div, X_dept, X_class, X_text])

print("Final feature matrix:", X.shape)
print("Target:", target.shape)
```

## Train / validation / test split

The split is performed manually with NumPy.

The normalization statistics are calculated **only from the training set**, preventing test-set information leakage.


```python
# Stratified split implemented with NumPy.
rng = np.random.default_rng(RANDOM_SEED)

train_idx = []
val_idx = []
test_idx = []

for cls in [0, 1]:
    idx = np.where(target == cls)[0]
    rng.shuffle(idx)
    n = len(idx)
    n_train = int(0.70*n)
    n_val = int(0.15*n)
    train_idx.extend(idx[:n_train])
    val_idx.extend(idx[n_train:n_train+n_val])
    test_idx.extend(idx[n_train+n_val:])

train_idx = np.array(train_idx)
val_idx = np.array(val_idx)
test_idx = np.array(test_idx)

rng.shuffle(train_idx)
rng.shuffle(val_idx)
rng.shuffle(test_idx)

X_train, X_val, X_test = X[train_idx], X[val_idx], X[test_idx]
y_train = target[train_idx].reshape(-1,1)
y_val = target[val_idx].reshape(-1,1)
y_test = target[test_idx].reshape(-1,1)

mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
std[std < 1e-8] = 1.0

X_train = (X_train - mean) / std
X_val = (X_val - mean) / std
X_test = (X_test - mean) / std

print("Train:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Test:", X_test.shape, y_test.shape)
```
